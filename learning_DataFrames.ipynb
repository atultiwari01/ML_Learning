```python
sold_by_age = ipl_auction_df.groupby('AGE')['SOLD PRICE'].mean().reset_index()
print(sold_by_age)
```

```python
sold_by_age_role = ipl_auction_df.groupby(['AGE','PLAYING ROLE'])['SOLD PRICE'].mean().reset_index()
print(sold_by_age_role)
```

```python
soldprice_comparison = sold_by_age_role.merge(sold_by_age,on='AGE',how='outer')
print(soldprice_comparison)
```
